```python
import random

print("Záhada Tajemného lesa")
print("Vítejte v místech, kde se všem tají dech a kde můžete prověřit svou odvahu a vědomosti.")
print("POZOR! Něco se pohybuje mezi stromy.")
print("Co v tu chvílí uděláš?")
rozhodnutí = input("1 = Schováš se, 2 = Utečeš: ")
while rozhodnutí != "1" and rozhodnutí != "2":
    print("Neplatná volba, zadej znovu!")
    rozhodnutí = input("1 = Schováš se, 2 = Utečeš: ")
if rozhodnutí == "1":
    print("Rozhodl ses schovat. Najdeš si místo za stromem a čekáš, až zvuky utichnou.")
    print("UF, po chvíli se zvuky ztiší a ty se cítíš v bezpečí.")
    print("Po dlouhých minutách vylézáš opatrně ze svého úkrytu.")
    print("Zjistíš, že to byl jen silný vítr.")
    print("Nečekaně ale po tobě skočí divoký vlk.")
    print("Umíráš. Tvoje hra končí!")
    quit
elif rozhodnutí == "2":
    print("Rozhodl ses utéct. Běžíš přes les a snažíš se najít cestu, kterou se dostaneš do bezpečí.")
    print("Po chvíli tě dohoní malý skřet, s kterým musíš bojovat.")
    print("Bojuješ s ním tak, že počítáš matematické příklady.")
    číslo1 = random.randint(1, 10)
    číslo2 = random.randint(1, 10)
    odpověd = input(f"Kolik je {číslo1} * {číslo2}? ")
    while not odpověd.isdigit():
        print("Neplatná odpověď! Zadej číslo.")
        odpověd = input(f"Kolik je {číslo1} * {číslo2}? ")
    if int(odpověd) == číslo1 * číslo2:
        print("Správně! Podařilo se ti zabít protivníka.")
        print("Pak jdeš dál.")
        print("Procházíš dál Tajemným lesem. Za zatáčkou se před tebou objeví věž.")
        věž = input("Chceš jít do věže? 1 = Ano, 2 = Ne")
        while věž != "1" and věž != "2":
            print("Neplatná volba, zadej znovu!")
            věž = input("Chceš jít do věže? 1 = Ano, 2 = Ne")
        if věž == "1":
            print("Vcházíš do věže a překvapuje tě jak je vysoká a kolik má schodů.")
            schody = input("Chceš jít nahoru nebo chceš zůstat dole? 1 = Jít nahoru, 2 = Zůstat dole")
            while schody != "1" and schody != "2":
                print("Neplatná volba, zadej znovu!")
                schody = input("Chceš jít nahoru nebo chceš zůstat dole? 1 = Jít nahoru, 2 = Zůstat dole")
            if schody == "1":
                print("Vydáváš se nahoru po schodech. Tvým úkolem je z hlavy sečíst označené schody.")
                schody_1 = random.randint(10, 100)
                schody_2 = random.randint(10, 100)
                odpověd_schody = input(f"Kolik je {schody_1} + {schody_2}? ")
                while not odpověd_schody.isdigit():
                    print("Neplatná odpověď! Zadej číslo.")
                    odpověd_schody = input(f"Kolik je {schody_1} + {schody_2}? ")
                if int(odpověd_schody) == schody_1 + schody_2:
                    print("Správně! Došel jsi nahoru.")
                    print("Shora se před tebou rozprostírá celý les. A tak vidíš i bezpečnou cestu ven.")
                    print("Musiš se vydat na jih.")
                    print("Jdeš a jdeš a jdeš. Najednou je jen kousek od tebe konec lesa.")
                    print("Doběhneš až k jeho okraji.")
                    print("Zhluboka se nadechneš se a pomyslíš si: Konečně jsem pryč.")
                    print("Gratuluji, odhalil jsi záhadu Tajemného lesa, protože jsi pochopil, že než se schovávat za stromem, je lepší být odvážný a postavit se svému strachu.")
                    quit
                else:
                    print("Škoda! Došel ti dech a zamotala se ti hlava.")
                    print("Skácel si se ze všech schodů dolů a tvůj život skončil.")
                    quit
            elif schody == "2":
                print("Rozhodl ses zůstat dole. Prohledáváš přízemí věže.")
                print("Vysoko nad oknem visí velký klíč, který rychle sundaváš.")
                print("Dál procházíš celou místnost a v jejím rohu nacházíš truhlu.")
                print("Bereš klíč do ruky a truhlu pomalu otevíráš.")
                print("Na jejím dně leží zlatý budík.")
                print("Zvedáš ho, a ručičky hodin ukazují 7:00.")
                print("Zničeho nic se budík rozdrnčí. Trhneš sebou a otevíráš oči dokořán.")
                print("Nejsi ve věži ani v Tajemném lese. Jsi u sebe doma a právě ti došlo, že šlo o tvůj další bláznivý sen!")
                quit
        elif věž == "2":
            print("Rozhodl ses nejít do věže, jdeš tedy dál do hlouby Tajemného lesa.")
            print("Najednou se před tebou v mlze objeví chata.")
            print("V chatě sedí starý pán, který údajně ví jak se dostat z lesa ven.")
            print("Nabízí ti pomoc.")
            pomoc = input("Necháš si od něj pomoct? 1 = Ano, 2 = Ne")
            while pomoc != "1" and pomoc != "2":
                print("Neplatná volba, zadej znovu!")
                pomoc = input("Necháš si od něj pomoct? 1 = Ano, 2 = Ne")
            if pomoc == "1":
                print("Rozhodl ses, že si necháš od starého pána poradit.")
                print("Řekne ti ale, že dnes už je na cestu pozdě a tak u něj máš přespat.")
                print("Jdeš tedy spát.")
                print("Druhý den se probouzíš a čekáš než se probudí i starý pán.")
                print("Mezitím se rozhlížíš po domě.")
                print("Za pár chvil se probouzí i děd.")
                print("Zeptá se tě jestli jsi připraven vydat se na cestu z lesa.")
                print("Souhlasíš, a tak vyrážíte na cestu.")
                print("Po dlouhých hodinách putování dorazíte k propasti.")
                print("Starý pán se zničeho nic promění v čaroděje a vší silou tě strčí přes okraj srázu.")
                print("Umíráš. Tvoje cesta Tajemným lesem končí.")
                quit
            elif pomoc == "2":
                print("Odmítáš pomoc starého pána.")
                print("Muž se tě snaží přemluvit.")
                přemluva = input("Podařilo se mu tě přemluvit? 1 = Ano, 2 = Ne")
                while přemluva != "1" and přemluva != "2":
                    print("Neplatná volba, zadej znovu!")
                    přemluva = input("Podařilo se mu tě přemluvit? 1 = Ano, 2 = Ne")
                if přemluva == "1":
                    print("Přemluvil tě!.")
                    print("Řekne ti ale, že dnes už je na cestu pozdě a tak u něj máš přespat.")
                    print("Jdeš tedy spát.")
                    print("Druhý den se probouzíš a čekáš než se probudí i starý pán.")
                    print("Mezitím se rozhlížíš po domě.")
                    print("Za pár chvil se probouzí i děd.")
                    print("Zeptá se tě jestli jsi připraven vydat se na cestu z lesa.")
                    print("Souhlasíš, a tak vyrážíte na cestu.")
                    print("Po dlouhých hodinách putování, dorazíte k propasti.")
                    print("Starý pán se zničeho nic promění v čaroděje a vší silou tě strčí přes okraj srázu.")
                    print("Umíráš. Tvoje cesta Tajemným lesem končí.")
                    quit
                elif přemluva == "2":
                    print("Nenecháš se přemluvit.")
                    print("Starý pán tě tedy vyhodí z chaty.")
                    print("Nevzdáváš se a procházíš dál Tajemným lesem.")
                    print("Po několika kilometrech přicházíš na rozcestí.")
                    rozcestí = input("Můžeš jít doleva nebo doprava. 1 = Doleva, 2 = Doprava")
                    while rozcestí != "1" and rozcestí != "2":
                        print("Neplatná volba, zadej znovu!")
                        rozcestí = input("Můžeš jít doleva nebo doprava. 1 = Doleva, 2 = Doprava")
                    if rozcestí == "1":
                        print("Jdeš doleva.")
                        print("Pomalu kráčíš lesem, dokud se neobjevíš před velikou jeskyní.")
                        print("V temném koutu jeskyně se ukrývá zamčená truhla.")
                        print("Musíš najít klíč.")
                        print("Prohledáváš každý kout dokud klíč nenajdeš.")
                        print("Otevíráš truhlu, na jejímž dně je jediná mince.")
                        print("Bereš ji do ruky a dáváš do kapsy.")
                        print("Jeskyně nemá druhý východ a tak se vracíš tudy kudy si do ní vstoupil.")
                        print("Jdeš dál a dál a dál Tajemným lesem.")
                        print("Najednou narazíš na štěrkovou cestu, po které se rozhodneš pokračovat. Po pár minutách chůze si všimneš autobusové zastávky.")
                        print("Vydáš se k ní zjistit jestli tudy jezdí autobusy.")
                        print("V tu chvíli ti ale dojde, že nevíš kolik je hodin, protože nemáš hodinky.")
                        print("Zkusíš to tedy odhadnout podle slunce na obloze. Tvůj tip je přibližně 15:00.")
                        print("Jízdní řád ukazuje, že by další autobus mohl jet už za 15 minut.")
                        print("Čekáš, čekáš a čekáš.")
                        print("Pomalu ztrácíš naději.")
                        print("Když v tu chvíli vyjíždí ze zatáčky starý autobus.")
                        print("Nastoupíš do něj, jenže řidič chce po tobě 20 Kč za jízdenku.")
                        print("Na chvíli se zarazíš, než ti dojde, že přesně tolik si našel na dně truhly.")
                        print("Gratuluji! V pořádku jsi prošel Tajemným lesem a dostal se do bezpečí.")
                        quit
                    elif rozcestí == "2":
                        print("Jdeš doprava.")
                        print("V dálce vidíš malé světlo.")
                        print("Jdeš k němu a spatříš posed.")
                        print("Na posedu sedí myslivce.")
                        print("Zamáváš na něj, myslivec si tě ale splete s divokým prasetem a zastřelí tě.")
                        print("Škoda! Už jsi byl jen kousek od cesty z lesa. Střela pro tebe ale byla smrtelná!")
                        quit
    else:
        print("Špatně! Protivník tě porazil.")
        print("Zkus si procvičit násobliku. Tvoje hra právě končí!")
        quit
```
